```python
# ============================================================
# Projeto de demonstração
# Disciplina: Rede Neural Aplicada a Segurança Cibernética
# Professor: Dr André Serrano
#            Departamento de Engenharia de Produção
#            Faculdade de Tecnologia da universidade de Brasília
# Aluno: Leandro Martins
# Objetivo: Testar o envio de um notebook do Google Colab
#           para um repositório no GitHub.
# ============================================================

# Dados fictícios de tentativas de login (logs)
tentativas_login = [
    {"usuario": "usuario01", "tentativas": 1, "horario": 10},
    {"usuario": "usuario02", "tentativas": 7, "horario": 3},
    {"usuario": "usuario03", "tentativas": 2, "horario": 14},
    {"usuario": "usuario04", "tentativas": 10, "horario": 2},
]

print("=== Análise de tentativas de login ===\n")

for login in tentativas_login:

    usuario = login["usuario"]
    tentativas = login["tentativas"]
    horario = login["horario"]

    # Regra simples para fins de demonstração
    if tentativas >= 5:
        classificacao = "⚠️ SUSPEITO"
    else:
        classificacao = "✅ NORMAL"

    print(f"Usuário: {usuario}")
    print(f"Tentativas: {tentativas}")
    print(f"Horário: {horario}:00")
    print(f"Classificação: {classificacao}")
    print("-" * 35)
```
